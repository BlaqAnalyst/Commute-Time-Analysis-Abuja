# Abuja Commute Time Analysis

Analysis of commute time patterns using the adjusted Abuja dataset.

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('Abuja_Commute_Dataset_Final_Adjusted.xlsx')
df.head()
```

```
df.info()
df.describe(include='all')
print(df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
```

```
df['Date']=pd.to_datetime(df['Date'],dayfirst=True,errors='coerce')
df['Fare_NGN']=pd.to_numeric(df['Fare_NGN'],errors='coerce')
df['Commute_Time_min']=pd.to_numeric(df['Commute_Time_min'],errors='coerce')
df=df.drop_duplicates()
df.head()
```

```
avg_time=df['Commute_Time_min'].mean()
avg_fare=df['Fare_NGN'].mean()
print('Average commute time:',avg_time)
print('Average fare:',avg_fare)

print(df.groupby('Transport_Mode')['Fare_NGN'].mean())
print(df.groupby('Traffic_Level')['Commute_Time_min'].mean())
```

```
df['Transport_Mode'].value_counts().plot(kind='bar')
plt.title('Transport Mode Frequency')
plt.tight_layout()
plt.show()

df.groupby('Transport_Mode')['Fare_NGN'].mean().plot(kind='bar')
plt.title('Average Fare by Transport Mode')
plt.tight_layout()
plt.show()

df.groupby('Traffic_Level')['Commute_Time_min'].mean().plot(kind='bar')
plt.title('Average Commute Time by Traffic Level')
plt.tight_layout()
plt.show()
```
